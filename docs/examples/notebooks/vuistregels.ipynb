De vuistregel Agent van 'Talk with the Engineer' heeft drie verschillende tools tot zijn beschikking. Het taalmodel kan deze zelf invullen, maar hier geven we per tool een voorbeeld van de mogelijke in en output.

## Voorbereidingen
In eerste instantie definieren we een context object.

```python
from pydantic_ai import format_as_xml
from setup_quarto import setup_dot_env

ctx = setup_dot_env()
```

## Vasthouden

In de trits, vasthouden-bergen-afvoeren bespreken we hier allereerst de tool voor 'vasthouden' (`hold_tool`)

```python
from talk_with_the_engineer.tools import hold_tool
from talk_with_the_engineer.models import HoldInput, HoldOutput

hold_input = HoldInput(
    area_ha=500, 
    time_of_year='summer', 
    return_period=10, 
    delta_h_available=0.3,
    percentage_open_water=5,
    discharge_capacity_mm_d=2.6
)

hold_output: HoldOutput = await hold_tool(ctx=ctx, input=hold_input)
print(format_as_xml(hold_output))
```

## Bergen

In de trits, vasthouden-bergen-afvoeren bespreken we hier allereerst de tool voor 'bergen' (`store_tool`)

```python
from talk_with_the_engineer.tools import store_tool
from talk_with_the_engineer.models import StoreInput, StoreOutput

store_input = StoreInput(
    area_ha=500, 
    time_of_year='summer', 
    return_period=10, 
    delta_h_available=0.3,
    percentage_open_water=5,
    discharge_capacity_mm_d=2.6
)

store_output: StoreOutput = await store_tool(ctx=ctx, input=store_input)
store_output.model_dump()
```

In de trits, vasthouden-bergen-afvoeren bespreken we hier allereerst de tool voor 'afvoeren' (`discharge_tool`)

```python
from talk_with_the_engineer.tools import discharge_tool
from talk_with_the_engineer.models import DischargeInput, DischargeOutput

discharge_input = DischargeInput(
    area_ha=500, 
    time_of_year='summer', 
    return_period=10, 
    delta_h_available=0.3,
    percentage_open_water=5,
    discharge_capacity_mm_d=2.6
)

discharge_output: DischargeOutput = await discharge_tool(ctx=ctx, input=discharge_input)
discharge_output.model_dump()
```
