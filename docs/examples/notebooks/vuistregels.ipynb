De vuistregel Agent van 'Talk with the Engineer' heeft drie verschillende tools tot zijn beschikking. Het taalmodel kan deze zelf invullen, maar hier geven we per tool een voorbeeld van de mogelijke in en output.

## Voorbereidingen
In eerste instantie definieren we een context object.

```python
from pydantic_ai import format_as_xml
from setup_quarto import setup_dot_env

ctx = setup_dot_env()
```

## Vasthouden

In de trits, vasthouden-bergen-afvoeren bespreken we hier allereerst de tool voor 'vasthouden' (`hold_tool`)

```python
from talk_with_the_engineer.tools import calculate_retention_given_h
from talk_with_the_engineer.models import RetentionInputGivenH, RetentionOutput


hold_input = RetentionInputGivenH(
    area_ha=4000,
    time_of_year="summer",
    return_period=10,
    delta_h_available=1.5,
    percentage_open_water=3.5,
    discharge_capacity_mm_d=2.6,
)

hold_output: RetentionOutput = await calculate_retention_given_h(
    ctx=ctx, input=hold_input
)
print(format_as_xml(hold_output))
```

## Bergen

In de trits, vasthouden-bergen-afvoeren bespreken we hier allereerst de tool voor 'bergen' (`store_tool`)

```python
from talk_with_the_engineer.tools import calculate_store_given_h
from talk_with_the_engineer.models import StoreInputGivenH, StoreOutput

store_input = StoreInputGivenH(
    area_ha=4000,
    time_of_year="summer",
    return_period=10,
    delta_h_available=1.5,
    percentage_open_water=3.5,
    discharge_capacity_mm_d=2.6,
)

store_output: StoreOutput = await calculate_store_given_h(ctx=ctx, input=store_input)
store_output.model_dump()
```

```python
from talk_with_the_engineer.tools import calculate_store_given_area
from talk_with_the_engineer.models import StoreInputGivenArea, StoreOutput

store_input = StoreInputGivenArea(
    area_ha=4000,
    time_of_year="summer",
    return_period=10,
    percentage_open_water=3.5,
    area_available_ha=20,
    discharge_capacity_mm_d=2.6,
)

store_output: StoreOutput = await calculate_store_given_area(ctx=ctx, input=store_input)
store_output.model_dump()
```

## Afvoeren

In de trits, vasthouden-bergen-afvoeren bespreken we hier allereerst de tool voor 'afvoeren' (`discharge_tool`)

```python
from talk_with_the_engineer.tools import calculate_discharge_given_h
from talk_with_the_engineer.models import DischargeInputGivenH, DischargeOutput

discharge_input = DischargeInputGivenH(
    area_ha=4000,
    time_of_year="summer",
    return_period=10,
    delta_h_available=1.5,
    percentage_open_water=3.5,
    discharge_capacity_mm_d=2.6,
)

discharge_output: DischargeOutput = await calculate_discharge_given_h(
    ctx=ctx, input=discharge_input
)
discharge_output.model_dump()
```
