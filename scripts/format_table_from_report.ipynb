Extremen gebiedsneerslag (mm) gedurende het gehele jaar en het winterhalfjaar voor duren van 2
t/m 8 dagen, die gemiddeld 2 maal per jaar tot gemiddeld eens per 1000 jaar worden
overschreden. <br>

reductiefactoren van Tabel 6 uit deel 4 STOWA, 2019).

```python
data_str = """"Duur (uren) 48 96 192 48 96 192
2 0.5 16.0 29.0 47.0 29.0 39.5 54.8
1 1 22.0 35.0 55.0 34.4 47.0 65.6
0.5 2 28.0 42.0 63.0 40.1 54.7 76.0
0.2 5 36.0 51.0 73.0 48.0 65.1 89.4
0.1 10 43.0 59.0 81.0 54.2 73.0 99.1
0.05 20 50.0 66.0 89.0 60.7 81.1 108.5
0.04 25 53.0 69.0 92.0 62.9 83.8 111.5
0.02 50 61.0 77.0 99.0 69.7 92.0 120.5
0.01 100 70.7 86.6 108.1 76.8 100.4 129.3
0.005 200 81.1 96.5 117.2 84.1 108.9 137.7
0.004 250 84.6 99.7 120.2 86.6 111.7 140.4
0.002 500 96.0 110.1 129.3 94.3 120.4 148.5
0.001 1000 108.2 120.9 138.4 102.4 129.2 156.4"""
```

```python
lines = data_str.strip().split("\n")
header = lines[0].split()
table_dict = {" ".join(header[0:2]).strip('"'): [int(x) for x in header[2:]]}

for line in lines[1:]:
    parts = line.split()
    key = parts[0]
    values = [
        float(x.replace("½", "0.5")) if x.replace(".", "", 1).isdigit() else x
        for x in parts[1:]
    ]
    table_dict[key] = values

for key in table_dict:
    if key != "Duur (uren)":
        table_dict[key] = table_dict[key][1:]
```

```python
rainfall_data_incl_reductie = table_dict
```

```python
summer = {}
winter = {}
for key in rainfall_data:
    summer[key] = rainfall_data[key][:3]
    winter[key] = rainfall_data[key][3:]
```

```python
df_winter = pd.DataFrame(winter)
df_winter = df_winter.T
df_winter.columns = [48.0, 96.0, 192.0]
df_winter.drop(index="Duur (uren)", inplace=True)
df_winter.index.name = "T"
df_winter.columns.name = "Duur (uren)"
df_winter.index = [1 / float(val) for val in df_winter.index]
```
