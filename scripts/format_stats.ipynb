```python
import pandas as pd
```

### Statistical rainfall patterns

```python
df_winter = pd.read_excel(
    "stats_data_in.xlsm", sheet_name="PatronenWinter", skiprows=[0], index_col=[0]
)
df_summer = pd.read_excel(
    "stats_data_in.xlsm", sheet_name="PatronenJaar", skiprows=[0], index_col=[0]
)
```

```python
df_winter.columns
```

```python
columns = ["p_0", "p1a", "p1b", "p1c", "p1d", "p2a", "p2b"]
df_winter_48 = df_winter[columns]
df_winter_96 = df_winter[
    ["p_0.1", "p1a.1", "p1b.1", "p1c.1", "p1d.1", "p2a.1", "p2b.1"]
]
df_winter_96.columns = columns
df_winter_192 = df_winter[
    ["p_0.2", "p1a.2", "p1b.2", "p1c.2", "p1d.2", "p2a.2", "p2b.2"]
]
df_winter_192.columns = columns


columns = ["p_0", "p1a", "p1b", "p1c", "p1d", "p2a", "p2b"]
df_summer_48 = df_summer[columns]
df_summer_96 = df_summer[
    ["p_0.1", "p1a.1", "p1b.1", "p1c.1", "p1d.1", "p2a.1", "p2b.1"]
]
df_summer_96.columns = columns
df_summer_192 = df_summer[
    ["p_0.2", "p1a.2", "p1b.2", "p1c.2", "p1d.2", "p2a.2", "p2b.2"]
]
df_summer_192.columns = columns
```

```python
df_out = {}
df_out["winter"] = {}
df_out["winter"][48] = df_winter_48.to_dict()
df_out["winter"][96] = df_winter_96.to_dict()
df_out["winter"][192] = df_winter_192.to_dict()

df_out["summer"] = {}
df_out["summer"][48] = df_summer_48.to_dict()
df_out["summer"][96] = df_summer_96.to_dict()
df_out["summer"][192] = df_summer_192.to_dict()


output = False
if output:
    import json

    with open("rainfall_patterns.json", "w") as f:
        json.dump(df_out, f)
```
