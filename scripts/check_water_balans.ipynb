```python
length_rainfall_days = int(96 / 24)
array_daily = get_rainfall_pattern("winter", 96, "p1c").to_numpy().reshape(8, 24)
array_daily = array_daily.sum(axis=1)
```

```python
volume_rain_total
```

```python
rainfall_per_day = array_daily[:length_rainfall_days] * volume_rain_total
```

```python
rainfall_per_day
```

```python
rainfall_d12 = sum(rainfall_per_day[:2])
```

```python
reduction_factor = (20 / sum(rainfall_per_day[:2]))
reduction_factor
```

```python
rainfall_per_day[:2] = [
            x * (1 - reduction_factor) for x in rainfall_per_day[:2]
        ]
```
